```python
import os
import pandas as pd

# Folder containing transcript files
transcript_folder = "../transcripts"

# Empty list to store data
data = []

# Read each transcript
for file in os.listdir(transcript_folder):
    if file.endswith(".txt"):

        file_path = os.path.join(transcript_folder, file)

        with open(file_path, "r", encoding="utf-8") as f:
            transcript = f.read()

        data.append({
            "Call_ID": file.replace(".txt", ""),
            "Transcript": transcript,
            "Word_Count": len(transcript.split())
        })

# Create DataFrame
df = pd.DataFrame(data)

# Create folder if it doesn't exist
os.makedirs("../structured_data", exist_ok=True)

# Save CSV
csv_path = "../structured_data/call_data.csv"
df.to_csv(csv_path, index=False)

print("✅ CSV created successfully!")
print(f"Location: {csv_path}")

# Display first 5 rows
df.head()
```
